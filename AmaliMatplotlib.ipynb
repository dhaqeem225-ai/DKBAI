#Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#Sediakan Data

```python
data = {
    "Nama": ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
    "Markah": [70, 85, 90, 65, 80]
}
df = pd.DataFrame(data)
```

#Pengiraan menggunakan NumPy

```python
purata = np.mean(df["Markah"])
tertinggi = np.max(df["Markah"])
terendah = np.min(df["Markah"])

print("Purata:", purata)
print("Markah tertinggi:", tertinggi)
print("Markah terendah:", terendah)
```

#Tambah status menggunakan Pandas

```python
df["Status"] = np.where(df["Markah"] >= 50, "Lulus", "Gagal")
print("\nDataframe Pelajar:")
print(df)
```

#Paparkan data menggunakan matplotlib

```python
plt.bar(df["Nama"], df["Markah"])
plt.axhline(
    y=purata,
    color="red",
    linestyle="--",
    label="Purata"
)
plt.title("Analisis Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.legend()
plt.show()
```
