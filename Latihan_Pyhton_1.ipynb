```python
print("--- 3. GANJIL ATAU GENAP ---")
nombor = int(input("Masukkan satu nombor bulat: "))

if nombor % 2 == 0:
    print(f"Nombor {nombor} ialah nombor GENAP.\n")
else:
    print(f"Nombor {nombor} ialah nombor GANJIL.\n")
```
