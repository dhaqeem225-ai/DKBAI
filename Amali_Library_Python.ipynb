#Import Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#Bina Array Markah

```python
markah = np.array([70, 85, 90, 65, 80])
print(markah)
```

#Purata markah

```python
purata = np.mean(markah)
print("Purata markah:", purata)
```

#Cari markah tertinggi


```python
tertinggi = np.max(markah)
print("Markah tertinggi:", tertinggi)
```

#Cari markah terendah

```python
terendah = np.min(markah)
print("Markah terendah:", terendah)
```

#Kira Jumlah Keseluruhan

```python
jumlah = np.sum(markah)
print("Jumlah semua markah:", jumlah)
```

#Tambah Markah Bonus

```python
markah_baru = markah + 5
print(markah_baru)
```

#AMALI PANDAS


```python

data = {
"Nama": ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
"Markah": [70, 85, 90, 65, 80]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
```

```python
print(df["Nama"])
```

```python
print(df["Markah"])
```

#Kira purata

```python
purata = df["Markah"].mean()
print("Purata markah:", purata)
```

#Tapis Pelajar Markah 80 keatas

```python
pelajar_cemerlang = df[df["Markah"] >= 80]
print(pelajar_cemerlang)
```

#Tambah lajur status

```python
df ["Status"]=np.where(df["Markah"] >= 65, "Lulus", "Gagal")

print(df)
```

#Tambah lajur gred

```python
def tentukan_gred(markah):
  if markah >= 80:
    return "A"
  elif markah >= 65:
    return "B"
  elif markah >= 50:
    return "C"
  else:
    return "D"

df["Gred"] = df["Markah"].apply(tentukan_gred)
print(df)
```

#Simpan data ke dalam fail CSV

```python
df.to_csv("markah_pelajar.csv", index=False)
print("Fail berjaya disimpan.")
```
